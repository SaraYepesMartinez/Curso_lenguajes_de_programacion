CALCULAR LA VPN PARA DARNOS CUENTA SI EL PROYECTO ES RENTABLE O NO LO ES, TENIENDO EN CONSIDERACION TODA LA INFORMACION SUMINISTRADA

```python
import pandas as pd

def calcular_vpn(flujos, tasa_descuento):
    vpn = 0
    for t in range(len(flujos)):
        vpn += flujos[t] / ((1 + tasa_descuento) ** t)
    return vpn

# Tasa de descuento 10%.

tasa = 0.10

def obtener_flujos(fila):
    return [fila['Inversion_Inicial']] + [fila[f'Año_{i}'] for i in range(1, 11)]

df['Flujos_Caja'] = df.apply(obtener_flujos, axis=1)
df['VPN'] = df.apply(lambda fila: calcular_vpn(fila['Flujos_Caja'], tasa), axis=1)

print(df[['Proyecto', 'VPN']].head())
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output
import pandas as pd
import matplotlib.pyplot as plt

# Leer el archivo CSV y realizar cálculos iniciales (se ejecutan al inicio)
try:
    df = pd.read_csv("/content/Flujos_de_caja.csv", sep=';', encoding='latin1')
    df.columns = ['Proyecto', 'Inversion_Inicial'] + [f'Año_{i}' for i in range(1, 11)]

    def calcular_vpn(flujos, tasa_descuento):
        vpn = 0
        for t in range(len(flujos)):
            vpn += flujos[t] / ((1 + tasa_descuento) ** t)
        return vpn

    tasa = 0.10
    def obtener_flujos(fila):
        return [fila['Inversion_Inicial']] + [fila[f'Año_{i}'] for i in range(1, 11)]
    df['Flujos_Caja'] = df.apply(obtener_flujos, axis=1)
    df['VPN'] = df.apply(lambda fila: calcular_vpn(fila['Flujos_Caja'], tasa), axis=1)

    def calcular_tir_manual(flujos, tolerancia=0.0001, max_iteraciones=100):
        tasa_bajo = -0.99
        tasa_alto = 10.0
        for _ in range(max_iteraciones):
            tasa_medio = (tasa_bajo + tasa_alto) / 2
            vpn = calcular_vpn(flujos, tasa_medio)
            if abs(vpn) < tolerancia:
                return tasa_medio
            elif vpn > 0:
                tasa_bajo = tasa_medio
            else:
                tasa_alto = tasa_medio
        return None
    df['TIR_Manual'] = df.apply(lambda fila: calcular_tir_manual(
        [fila['Inversion_Inicial']] + [fila[f'Año_{i}'] for i in range(1, 11)]
    ), axis=1)

    # Funciones para cada botón
    def mostrar_datos(b):
        clear_output(wait=True)
        display(menu_box)
        print("Primeras filas del DataFrame:")
        display(df.head())

    def mostrar_informacion(b):
        clear_output(wait=True)
        display(menu_box)
        print("Información general del DataFrame:")
        print(df.info())
        print("\nValores nulos por columna:")
        print(df.isnull().sum())
        print("\nResumen estadístico:")
        print(df.describe())

    def calcular_vpn_output(b):
        clear_output(wait=True)
        display(menu_box)
        print("Valor Presente Neto (VPN) por Proyecto:")
        display(df[['Proyecto', 'VPN']].head())

    def mostrar_rentabilidad(b):
        clear_output(wait=True)
        display(menu_box)
        print("Valor Presente Neto (VPN) y Tasa Interna de Retorno (TIR) Manual por Proyecto:")
        display(df[['Proyecto', 'VPN', 'TIR_Manual']].head())

    def mostrar_grafico(b):
        clear_output(wait=True)
        display(menu_box)
        plt.figure(figsize=(12, 5))
        plt.bar(df['Proyecto'], df['VPN'], color='purple')
        plt.axhline(0, color='red', linestyle='--')
        plt.title('Valor Presente Neto (VPN) por Proyecto')
        plt.ylabel('VPN')
        plt.xlabel('Proyecto')
        plt.xticks(rotation=45, ha='right')
        plt.tight_layout()
        plt.show()

    # Crear los botones con los nombres solicitados
    boton_datos = widgets.Button(description="DATOS")
    boton_informacion = widgets.Button(description="INFORMACIÓN")
    boton_calcular_vpn = widgets.Button(description="CALCULO VPN")
    boton_rentabilidad = widgets.Button(description="RENTABILIDAD")
    boton_grafico = widgets.Button(description="GRAFICO")

    # Asignar funciones a los botones
    boton_datos.on_click(mostrar_datos)
    boton_informacion.on_click(mostrar_informacion)
    boton_calcular_vpn.on_click(calcular_vpn_output)
    boton_rentabilidad.on_click(mostrar_rentabilidad)
    boton_grafico.on_click(mostrar_grafico)

    # Organizar los botones en un menú
    menu_box = widgets.HBox([boton_datos, boton_informacion, boton_calcular_vpn, boton_rentabilidad, boton_grafico])

    # Mostrar el menú
    display(menu_box)

except FileNotFoundError:
    print("Error: El archivo 'Flujos_de_caja.csv' no se encontró en la carpeta '/content/'.")
except Exception as e:
    print(f"Ocurrió un error: {e}")
```
